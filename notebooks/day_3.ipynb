```python
import numpy as np
import matplotlib.pyplot as plt

def loss(w):
    return (w - 3) ** 2

def gradient(w):
    return 2 * (w - 3)
```

```python
w = 0.0              # initial weight
lr = 0.1             # learning rate

new_w = w - lr * gradient(w)
new_w
```

```python
w = 0.0
lr = 0.1
history = []

for step in range(20):
    history.append(w)
    w = w - lr * gradient(w)

history
```

```python
w_values = np.linspace(-2, 6, 100)
loss_values = loss(w_values)

plt.plot(w_values, loss_values, label="Loss Function")
plt.scatter(history, loss(np.array(history)), color="red", label="GD Steps")
plt.xlabel("Weight (w)")
plt.ylabel("Loss")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

X = np.random.rand(100, 1)     # 100 data points
y = 2 * X + 1 + 0.1 * np.random.randn(100, 1)
```

```python
plt.scatter(X, y)
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
w = 0.0   # weight
b = 0.0   # bias
```

```python
def mse_loss(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)
```

```python
def compute_gradients(X, y, w, b):
    n = len(X)
    y_pred = w * X + b
    
    dw = (-2 / n) * np.sum(X * (y - y_pred))
    db = (-2 / n) * np.sum(y - y_pred)
    
    return dw, db
```

```python
lr = 0.1
epochs = 100

losses = []

for epoch in range(epochs):
    y_pred = w * X + b
    loss = mse_loss(y, y_pred)
    losses.append(loss)
    
    dw, db = compute_gradients(X, y, w, b)
    
    w = w - lr * dw
    b = b - lr * db
```

```python
w, b
```

```python
plt.scatter(X, y)
plt.plot(X, w * X + b, color="red")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.show()
```
